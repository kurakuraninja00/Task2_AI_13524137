# 01 — Exploratory Data Analysis
EDA untuk Loan Approval Classification dataset.

```python
import sys
sys.path.insert(0, '..')

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from src.data import load_train, load_test
from src.utils import set_seed

set_seed(42)
plt.style.use('seaborn-v0_8-whitegrid')
```

```python
train_df = load_train()
test_df = load_test()
print(f'Train shape: {train_df.shape}')
print(f'Test shape: {test_df.shape}')
train_df.head()
```

```python
# Distribusi target
fig, ax = plt.subplots(figsize=(6, 4))
train_df['loan_status'].value_counts().plot(kind='bar', color=['#e74c3c', '#2ecc71'], ax=ax)
ax.set_title('Distribusi loan_status', fontweight='bold')
ax.set_ylabel('Count')
plt.tight_layout()
plt.show()
print(train_df['loan_status'].value_counts())
```

```python
# Statistik deskriptif
train_df.describe()
```

```python
# Outlier: person_age & person_emp_exp
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
train_df.boxplot(column='person_age', ax=axes[0])
axes[0].set_title('person_age Boxplot')
train_df.boxplot(column='person_emp_exp', ax=axes[1])
axes[1].set_title('person_emp_exp Boxplot')
plt.tight_layout()
plt.show()

print('person_age > 90:', (train_df['person_age'] > 90).sum())
print('person_emp_exp > 50:', (train_df['person_emp_exp'] > 50).sum())
```

```python
# Crosstab previous_loan_defaults_on_file vs loan_status
ct = pd.crosstab(train_df['previous_loan_defaults_on_file'], train_df['loan_status'])
print(ct)

fig, ax = plt.subplots(figsize=(8, 5))
ct.plot(kind='bar', stacked=True, color=['#e74c3c', '#2ecc71'], ax=ax)
ax.set_title('previous_loan_defaults_on_file vs loan_status', fontweight='bold')
ax.set_ylabel('Count')
plt.tight_layout()
plt.show()
```

```python
# Distribusi person_income (long tail)
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
train_df['person_income'].hist(bins=100, ax=axes[0])
axes[0].set_title('person_income Distribution')
np.log1p(train_df['person_income']).hist(bins=100, ax=axes[1])
axes[1].set_title('log1p(person_income) Distribution')
plt.tight_layout()
plt.show()
```
